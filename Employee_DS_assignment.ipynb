**TASK 2: Visualization**

       We will create the following plots:
        A scatter plot of employee age ald employee salary, healthy_eating and active_lifestyle, health_eating and salary
        A count plot of groups
        A histogram of employee salary

```python
#create sctatter plot

plt.figure(figsize = (5,5))

sns.scatterplot(data = df,x = 'age', y = 'salary', color = 'green',edgecolor = 'linen',
                 alpha = 0.7)

plt.title("Salary by age")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```

```python
plt.figure(figsize = (10,5))

sns.scatterplot(data = df,x = 'healthy_eating', y = 'active_lifestyle', color  = 'red',
                edgecolor = 'linen', alpha = 0.7)

plt.title("Healthy Eating v/s Active Lifestyle")
plt.xlabel("Healthy Eating")
plt.ylabel("Active Lifestyle")
plt.show()
```

```python
plt.figure(figsize = (10,5))

sns.scatterplot(data = df,x = 'healthy_eating', y = 'salary', color  = 'blue',
                edgecolor = 'linen', alpha = 0.7)

plt.title("Healthy Eating v/s Salary")
plt.xlabel("Healthy Eating")
plt.ylabel("Salary")
plt.show()
```

```python
plt.figure(figsize = (10,5))

sns.countplot(data = df,x = 'groups',edgecolor = 'linen',
                 alpha = 0.7)

plt.title("Count Plot of Groups")
plt.xlabel("Groups")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize = (10,5))

sns.histplot(data = df,x = 'salary', color = 'orange',edgecolor = 'linen',
                 alpha = 0.7, bins = 20)

plt.title("Histogram of Salary")
plt.xlabel("Salary")
plt.ylabel("Count")

plt.show()
```

**Observations form Task 2**

  Generally, as the employee's age increase, there is an increase in the employee salary
  Generally, as the employee's healthy_eating habit increase, there is an increase in the employee salary as well
  However, the employee salary values are more spread out
  An employee of healthy_eating greater than 8 can be taken as 8 only
  An employee salary less than J000 can be considered as a lower salary
