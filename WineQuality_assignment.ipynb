```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("/content/Wine Quality Dataset.csv")
```

```python
data.head()
```

```python
data.shape
```

```python
data.columns
```

```python
data.info()
```

```python
plt.figure(figsize = (10,5))

sns.histplot(data = data, x = 'fixed acidity', color = 'orange',
             edgecolor= 'linen', alpha = 0.7, bins = 20)

plt.title("Histogram of fixed acidity")
plt.xlabel("Fixed Acidity")
plt.ylabel("Count")
plt.show()
```

```python
round(data["fixed acidity"].mean(),2)
```

```python
data["fixed acidity"].median()
```

```python
data["fixed acidity"].mode()
```

```python
plt.figure(figsize = (10,8))

sns.histplot(data = data, x = 'fixed acidity', color = 'orange',
             edgecolor= 'linen', alpha = 0.7, bins = 10)

plt.title("Histogram of fixed acidity")
plt.xlabel("Fixed Acidity")
plt.ylabel("Count")
plt.vlines(data['fixed acidity'].mean(), ymin = 0, ymax = 4000, colors = 'blue', label = 'Mean')
plt.vlines(data['fixed acidity'].median(), ymin = 0, ymax = 4000, colors = 'green', label = 'Median')
plt.show()
```

```python
plt.figure(figsize = (10,5))

sns.histplot(data = data, x = 'volatile acidity', color = 'blue',
             edgecolor= 'linen', alpha = 0.7, bins = 10)

plt.title("Histogram of volatile acidity")
plt.xlabel("volatile Acidity")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize = (10,8))

sns.distplot(data['volatile acidity'], color = 'green')

plt.title("Distplot of Volatile Acidity")
plt.xlabel("Volatile Acidity")
plt.ylabel("Density")
plt.show()
```

```python
round(data['volatile acidity'].skew(),2)
```

```python
plt.figure(figsize = (10,8))

sns.histplot(data = data, x = 'volatile acidity', color = 'orange',
             edgecolor= 'linen', alpha = 0.7, bins = 10)

plt.title("Histogram of volatile acidity")
plt.xlabel("Volatile Acidity")
plt.ylabel("Count")
plt.vlines(data['volatile acidity'].mean(), ymin = 0, ymax = 4000, colors = 'blue', label = 'Mean')
plt.vlines(data['volatile acidity'].median(), ymin = 0, ymax = 4000, colors = 'green', label = 'Median')
plt.show()
```

```python
plt.figure(figsize = (10,8))

sns.histplot(data = data, x = 'citric acid', color = 'orange',
             edgecolor= 'linen', alpha = 0.7, bins = 10)

plt.title("Histogram of Citric acidity")
plt.xlabel("Citric Acid")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize = (10,8))

sns.histplot(data = data, x = 'citric acid', color = 'orange',
             edgecolor= 'linen', alpha = 0.7, bins = 10)

plt.title("Histogram of volatile acidity")
plt.xlabel("Citric Acidity")
plt.ylabel("Count")
plt.vlines(data['citric acid'].mean(), ymin = 0, ymax = 4000, colors = 'blue', label = 'Mean')
plt.vlines(data['citric acid'].median(), ymin = 0, ymax = 4000, colors = 'green', label = 'Median')
plt.show()
```

```python
plt.figure(figsize = (10,8))

sns.distplot(data["citric acid"], color = 'green')

plt.title("Distplot of Citric Acid")
plt.xlabel("Citric Acid")
plt.ylabel("Density")
plt.show()
```

```python
plt.figure(figsize = (11,6))

sns.countplot(data= data , x= 'quality')

plt.title("Countplot of quality")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```
